# Exploratory Data Analysis — Bank Loan Approval Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../dataset/loan_approval_dataset.csv')
df.columns = [c.strip() for c in df.columns]
df.head()
```

```python
df.info()
df.describe()
```

## Target class balance

```python
df['loan_status'].str.strip().value_counts().plot(kind='bar', title='Loan Status Distribution')
plt.show()
```

## CIBIL score vs loan status
Known to be a very strong predictor in this dataset — worth checking for near-perfect separability.

```python
sns.boxplot(x=df['loan_status'].str.strip(), y=df['cibil_score'])
plt.title('CIBIL Score by Loan Status')
plt.show()
```

## Correlation heatmap (numeric features)

```python
numeric_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()
```

## Income vs loan amount, colored by approval

```python
sns.scatterplot(data=df, x='income_annum', y='loan_amount', hue=df['loan_status'].str.strip(), alpha=0.5)
plt.title('Income vs Loan Amount by Approval Status')
plt.show()
```
